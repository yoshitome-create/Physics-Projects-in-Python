```python
import numpy as np
import matplotlib.pyplot as plt

```

```python
# Parameters
N = 1000
t = 1.0
mu = 0.0
g = 1.5

# Momentum grid
k = np.linspace(-np.pi, np.pi, N)

# Tight-binding dispersion
xi = -2 * t * np.cos(k) - mu

# Initial gap
Delta = 0.5

# Self-consistency loop
tol = 1e-8
max_iter = 1000

```

```python
for iteration in range(max_iter):

    E = np.sqrt(xi**2 + Delta**2)

    Delta_new = g * np.mean(Delta / (2 * E))

    if abs(Delta_new - Delta) < tol:
        break

    Delta = Delta_new

print("Converged gap =", Delta)
```

```python
# Plot quasiparticle spectrum
plt.plot(k, E)
plt.xlabel("k")
plt.ylabel("E_k")
plt.title("BCS quasiparticle spectrum")
plt.show()
```
